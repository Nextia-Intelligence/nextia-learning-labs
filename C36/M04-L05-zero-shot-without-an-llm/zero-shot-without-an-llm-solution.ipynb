## Setup: helpers

The next cell loads pandas as `pd`, numpy as `np` and the course helper as `rt`, reads the three parts of the inbox (`train`, `validation`, `test`), and defines the check helpers. They tell you if your answer is right, without showing the answer:

- `expect(what, yours, expected)` compares one of your numbers with the expected one, and says "too high" or "too low".
- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. The route names, as short phrases

A zero-shot model never saw a Larkfield label. It reads a short **phrase** for each route instead. These phrases are the labelling guidelines in a few words.

```python
import time, torch
torch.set_num_threads(4)
PHRASES = {
    "orders": "changing or cancelling an order",
    "delivery": "a late or missing delivery",
    "returns_refunds": "a return, a refund or a damaged item",
    "billing": "a charge, a payment or an invoice",
    "product": "a question about a product",
    "account": "a customer account or login",
    "needs_person": "a safety problem, a legal threat or a complaint to a manager",
}
TEMPLATE = "This message is about {}."
message = test.loc[test["message_id"] == "M-100597", "text"].item()
print(message)
```

## 2. NLI: one check per route

An **NLI** model (natural language inference) reads two sentences and says if the first one supports the second. Here, the first sentence is the message, and the second is a guess: "This message is about a late or missing delivery." The model checks seven guesses, one per route, and the route with the best score wins.

> **Predict.** The message is about tracking that has not moved. Which route will the NLI model choose?

```python
from transformers import pipeline
nli = pipeline("zero-shot-classification", model="MoritzLaurer/mDeBERTa-v3-base-xnli-multilingual-nli-2mil7", revision="b5113eb38ab63efdd7f280f8c144ea8b13f978ce", device=-1)
t0 = time.time()
out = nli(message, candidate_labels=list(PHRASES.values()), hypothesis_template=TEMPLATE)
ms = (time.time() - t0) * 1000
route_of = {v: k for k, v in PHRASES.items()}
for phrase, s in zip(out["labels"], out["scores"]):
    print(f"{route_of[phrase]:16} {s:.3f}   {TEMPLATE.format(phrase)}")
print(f"{ms:.0f} ms for 7 checks")
```

> **Check.** The NLI model puts **product** first. The message names a product (solar lights), and the guess "a question about a product" matched it better than the delivery guess. The time is around a second or more: 7 full passes of a 280-million-parameter model.

## 3. GLiClass: all routes in one pass

A **GLiClass** model reads the message and all seven phrases together, in one pass, and gives each phrase a score.

```python
from gliclass import GLiClassModel, ZeroShotClassificationPipeline
from transformers import AutoTokenizer
gmodel = GLiClassModel.from_pretrained("knowledgator/gliclass-multilang-edge", revision="d16c08ef70547514081952104e6fc3d190d9ee39").float()
gtok = AutoTokenizer.from_pretrained("knowledgator/gliclass-multilang-edge", revision="d16c08ef70547514081952104e6fc3d190d9ee39")
gli = ZeroShotClassificationPipeline(gmodel, gtok, classification_type="multi-label", device="cpu")

def gli_scores(texts, phrases):
    """One row per message, one column per phrase; the scores divided by their sum."""
    rows = []
    for text in texts:
        rows.append({x["label"]: x["score"] for x in gli(text, list(phrases), threshold=0.0)[0]})
    df = pd.DataFrame(rows).reindex(columns=list(phrases)).fillna(0.0)
    return df.div(df.sum(axis=1), axis=0)

gli_scores([message], PHRASES.values())  # warm-up
t0 = time.time()
s = gli_scores([message], PHRASES.values()).iloc[0]
print(pd.Series(s.to_numpy(), index=list(PHRASES)).round(3).sort_values(ascending=False))
print(f"{(time.time() - t0) * 1000:.0f} ms")
```

> **Check.** GLiClass puts **delivery** first, in a few tens of milliseconds.

## 4. Both models on the test messages

GLiClass is fast enough to run on all 2,000 test messages here (about a minute). NLI would take more than an hour, so we replay the course's recorded NLI predictions.

```python
t0 = time.time()
G = gli_scores(test["text"], PHRASES.values()).set_axis(list(PHRASES), axis=1)
print(f"GLiClass: {time.time() - t0:.0f} seconds")
nli_pred, nli_meta = rt.predictions("nli-zero-shot", test)
rt.show([rt.score("GLiClass zero-shot", test, G.idxmax(axis=1)),
         rt.score("NLI zero-shot (recorded)", test, nli_pred["prediction"])])
```

> **Check.** GLiClass: accuracy 0.516 (French 0.508). NLI: 0.329. Both cost far more per 1,000 messages than a person who reads everything (1,500 dollars): a wrong route is worse than no route.

## 5. Your turn: a new route without retraining

In the launch month, a new product arrives: the Grow Hub, a smart watering device. Chloé labelled 250 launch messages; 80 are about setting up the device (route `devices`). No model was trained on them. With a zero-shot model, a new route is one more phrase.

> **Your turn.** Score the 250 messages with GLiClass, with the 7 phrases plus the new phrase. Compute the **recall** of `devices` (the share of the 80 device messages that get `devices`) in `dev_recall`, and the **precision** (the share of messages predicted `devices` that really are) in `dev_precision`. Round both to 3 decimals.

```python
launch = pd.read_csv("data/launch_labels_sample.csv")
texts = pd.read_csv("data/launch_month.csv").set_index("message_id")["text"]
launch["text"] = texts.loc[launch["message_id"]].to_numpy()
NEW = dict(PHRASES, devices="setting up a smart watering device and its app")
pred = gli_scores(launch["text"], NEW.values()).set_axis(list(NEW), axis=1).idxmax(axis=1).to_numpy()
is_dev = launch["route"].to_numpy() == "devices"
dev_recall = round(float((pred[is_dev] == "devices").mean()), 3)
dev_precision = round(float((is_dev[pred == "devices"]).mean()), 3)
print(dev_recall, dev_precision)
```

```python
expect('dev_recall', dev_recall, 0.388, tolerance=0.0005)
```

```python
expect('dev_precision', dev_precision, 0.838, tolerance=0.0005)
```

GLiClass finds about 4 in 10 device messages, and most of the messages it calls `devices` are right. That is useful on day one, before anyone labels anything. It is not a router.
