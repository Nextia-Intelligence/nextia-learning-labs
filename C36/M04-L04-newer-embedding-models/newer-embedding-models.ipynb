## Setup: helpers

The next cell loads pandas as `pd`, numpy as `np` and the course helper as `rt`, reads the three parts of the inbox (`train`, `validation`, `test`), and defines the check helpers. They tell you if your answer is right, without showing the answer:

- `expect(what, yours, expected)` compares one of your numbers with the expected one, and says "too high" or "too low".
- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. Two models, side by side

The **older** model is the one of the last lessons (2021). The **newer** one was published in 2026. Both are multilingual, both give 384 numbers per message, and both have the Apache-2.0 licence.

> **Predict.** The newer model is five years younger. Will it route Larkfield's messages better?

```python
import time, torch
from sentence_transformers import SentenceTransformer
torch.set_num_threads(4)
torch.manual_seed(0)
MODELS = {'older (2021)': ('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2', 'e8f8c211226b894fcb81acc59f3b34ba3efd5f42'), 'newer (2026)': ('ibm-granite/granite-embedding-97m-multilingual-r2', '835ad14087e140460703cf0fae09f97d469d65c2')}
models = {}
for key, (name, rev) in MODELS.items():
    t0 = time.time()
    models[key] = SentenceTransformer(name, revision=rev, device="cpu")
    params = sum(p.numel() for p in models[key].parameters())
    dims = models[key].get_sentence_embedding_dimension()
    print(f"{key}: {name}, {params:,} parameters, {dims} numbers, loaded in {time.time() - t0:.0f} s")
```

> **Check.** The older model has 117,653,760 parameters, the newer 97,441,152. Both give 384 numbers.

## 2. Speed, one message at a time

```python
for key, m in models.items():
    m.encode(["warm-up"])
    t0 = time.time()
    for text in test["text"].head(50):
        m.encode([text], normalize_embeddings=True)
    print(f"{key}: {(time.time() - t0) * 20:.1f} ms per message")
```

> **Check.** On our laptop, the older model took about 7 ms per message and the newer about 27 ms. Your numbers depend on your computer.

When it loaded the newer model, the library printed a warning: it loaded the model in **bfloat16**, 16-bit numbers, which can be slow on a CPU. The next cell loads it again in 32-bit numbers and measures again.

```python
m32 = SentenceTransformer("ibm-granite/granite-embedding-97m-multilingual-r2", revision="835ad14087e140460703cf0fae09f97d469d65c2", device="cpu",
                          model_kwargs={"dtype": torch.float32})
m32.encode(["warm-up"])
t0 = time.time()
for text in test["text"].head(50):
    m32.encode([text], normalize_embeddings=True)
print(f"newer (2026), 32-bit: {(time.time() - t0) * 20:.1f} ms per message")
del m32
```

> **Check.** In 32-bit numbers, the newer model takes about 7 ms per message on our laptop, as fast as the older one. The slow part was a loading setting, not the model. Read the warnings, and measure the way you will run.

## 5. Your own small test

Here is the habit to keep: a quick test of any new model on your own data. We use 100 labelled messages per route for training, and all 2,000 test messages. It takes about two minutes, most of it for the newer model.

```python
from sklearn.linear_model import LogisticRegression
small = train.groupby("route").sample(100, random_state=36)
for key, m in models.items():
    t0 = time.time()
    e_small = m.encode(small["text"].tolist(), batch_size=64, normalize_embeddings=True)
    e_test = m.encode(test["text"].tolist(), batch_size=64, normalize_embeddings=True)
    clf = LogisticRegression(C=10, max_iter=3000).fit(e_small, small["route"])
    row = rt.score(key, test, clf.predict(e_test))
    print(f"{key}: accuracy {row['accuracy']}, English {row['accuracy_en']}, French {row['accuracy_fr']} "
          f"({time.time() - t0:.0f} s)")
```

> **Check.** With 700 labels, the older model reaches about 0.836 and the newer about 0.859. A surprise: with 100 labels per route, the **newer** model wins. With all 6,000 labels (section 3), the older one wins.

So the answer depends on how many labels you train with. Test a new model with the amount of data that you will really use, and check that the order holds on more than one size.
