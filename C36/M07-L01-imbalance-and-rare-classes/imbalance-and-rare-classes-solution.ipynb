## Setup: helpers

The next cell loads pandas as `pd`, numpy as `np` and the course helper as `rt`, reads the three parts of the inbox (`train`, `validation`, `test`), and defines the check helpers. They tell you if your answer is right, without showing the answer:

- `expect(what, yours, expected)` compares one of your numbers with the expected one, and says "too high" or "too low".
- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. The rare class

An **account takeover** is when someone else uses a customer's account: a changed password or email, orders that the customer did not make. It is urgent, and a missed one costs 200 dollars in the cost table. It is also rare.

```python
ato = lambda df: (df["topic"] == "account_takeover").to_numpy()
print("train", ato(train).sum(), "| validation", ato(validation).sum(), "| test", ato(test).sum())
print(train.loc[ato(train), "text"].head(3).to_string())
```

> **Check.** You should see 44 in train, 12 in validation and 12 in test. With so few test messages, one message more or less changes the recall by several points: always give the count next to a rare class's recall.

## 2. Accuracy hides the rare class

The next cell trains one TF-IDF model on the **topics** (27 of them) and compares the accuracy with the **macro-F1**: the F1 of each topic, averaged with the same weight for every topic.

> **Predict.** Will the takeover topic have a recall near the overall accuracy?

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.pipeline import FeatureUnion, make_pipeline

def features():
    return FeatureUnion([("words", TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode")),
                         ("chars", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=3, sublinear_tf=True))])

topic_model = make_pipeline(features(), LogisticRegression(C=10, max_iter=3000)).fit(train["text"], train["topic"])
tp = topic_model.predict(test["text"])
print(f"accuracy {(tp == test['topic']).mean():.3f}, macro-F1 {f1_score(test['topic'], tp, average='macro'):.3f}")
print("takeover recall:", (tp[ato(test)] == "account_takeover").mean().round(3))
```

> **Check.** You should see accuracy 0.954 and macro-F1 0.955, and a takeover recall of 0.833: 10 of the 12 test takeovers. Here the topic model does well; the next cell shows a model whose only job is the takeover flag, and what the rare class does to it.

## 3. A takeover flag, three ways

Grace wants a **takeover flag**: a yes/no model that marks a possible account takeover, so the account team sees it first. We train it three ways:

- **default**: the rare class counts like any message;
- **class weights**: each takeover message counts as much as many normal ones (`class_weight="balanced"`);
- **resampling**: the takeover messages appear 10 times in the training data.

```python
y_train = ato(train)
variants = {}
for name, weight, copies in [("default", None, 1), ("class weights", "balanced", 1), ("resampling x10", None, 10)]:
    data = pd.concat([train] + [train[y_train]] * (copies - 1), ignore_index=True)
    m = make_pipeline(features(), LogisticRegression(C=10, max_iter=3000, class_weight=weight))
    m.fit(data["text"], ato(data))
    p = m.predict_proba(test["text"])[:, 1]
    flag = p >= 0.5
    variants[name] = {"recall": flag[ato(test)].mean().round(3), "flagged": int(flag.sum()),
                      "precision": round(ato(test)[flag].mean(), 3) if flag.any() else None}
    if name == "default":
        flag_model = m
pd.DataFrame(variants).T
```

> **Check.** The default flag has recall 0.417. With class weights, recall is 0.917 and 11 test messages are flagged.

## 4. Your turn: a threshold of its own

Class weights move the threshold for you. You can also keep the default model and choose its threshold yourself, **on validation**: the highest threshold that reaches the target recall. Grace's target: at least 90% of takeovers on validation.

> **Your turn.** With `flag_model`, find the highest threshold from 0.50 down to 0.01 (in steps of 0.01) whose recall on the **validation** takeovers is at least 0.9. Put it in `threshold`. Then the next cell reports it on test.

```python
p_val = flag_model.predict_proba(validation["text"])[:, 1]
for threshold in np.round(np.arange(0.50, 0.0, -0.01), 2):
    if (p_val >= threshold)[ato(validation)].mean() >= 0.9:
        break
threshold = float(threshold)
print(threshold)
```

> **Check.** On test, you should see recall 0.833 with 10 messages flagged.

A lower threshold sends more messages to the account team first. Here it flagged no other message, but on validation the same threshold was chosen from only 12 takeovers: on test it found 10 of 12, not 90%. With so few examples, a recall target is a hope, not a promise; say so next to the number. Each extra flag costs a little time; each missed takeover costs 200 dollars. That trade is the reason for a threshold of its own.
