- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. A bag of words, by hand

A **bag of words** counts each word of a message and forgets the order. Here are three tiny messages:

1. `where is my parcel`
2. `my parcel is late`
3. `refund my order`

The **vocabulary** is every word that appears: `is, late, my, order, parcel, refund, where`. Each message becomes a row of counts, one column per word.

> **Predict.** Which column is 1 in all three rows? Then run the cell.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

tiny = ["where is my parcel", "my parcel is late", "refund my order"]
bag = CountVectorizer()
counts = bag.fit_transform(tiny)
pd.DataFrame(counts.toarray(), columns=bag.get_feature_names_out(), index=["msg 1", "msg 2", "msg 3"])
```

> **Check.** You should see 3 rows and 7 columns. The column `my` is 1 in every row.

A word that is in every message, like `my`, does not help to tell messages apart. **TF-IDF** (term frequency × inverse document frequency) gives such words a small weight: the weight of a word goes down when it appears in many messages. Each row is then scaled to length 1.

```python
tfidf = TfidfVectorizer()
weights = tfidf.fit_transform(tiny)
pd.DataFrame(weights.toarray(), columns=tfidf.get_feature_names_out(), index=["msg 1", "msg 2", "msg 3"]).round(3)
```

> **Check.** In message 1, `where` has the weight 0.632 and `my` only 0.373.

## 2. One Larkfield message as numbers

Now the same on the 6,000 training messages. The vectorizer keeps the words and word pairs that appear in at least 2 messages (`min_df=2`), and removes accents so that `livré` and `livre` count as one word.

```python
words = TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode")
X_train = words.fit_transform(train["text"])
print(X_train.shape[0], "messages,", X_train.shape[1], "columns (words and word pairs)")
one = test.iloc[0]
v = words.transform([one["text"]])
names = words.get_feature_names_out()
print(one["text"])
print(sorted(((names[i], round(w, 3)) for i, w in zip(v.indices, v.data)), key=lambda x: -x[1])[:10])
```

> **Check.** You should see 6000 messages and 14076 columns. Most of the columns are 0 for one message: it has only its own words.

## 3. A first model: logistic regression on TF-IDF

The next cell trains logistic regression on the word features and scores it on the test messages with the course's columns.

> **Predict.** Will the model be better on English or on French messages? Why?

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

word_model = make_pipeline(TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode"),
                           LogisticRegression(C=10, max_iter=3000))
word_model.fit(train["text"], train["route"])
rows = [rt.score("TF-IDF words + logistic regression", test, word_model.predict(test["text"]))]
rt.show(rows)
```

> **Check.** You should see accuracy 0.968, English 0.973 and French 0.956.

## 4. Character n-grams

A **character n-gram** is a group of n letters inside a word: `parcel` has the 3-grams `par`, `arc`, `rce`, `cel`. Character n-grams help with typos (`delivred`), with word endings (`remboursé`, `remboursement`) and with short messages. The next cell adds character n-grams of 2 to 5 letters to the word features.

```python
from sklearn.pipeline import FeatureUnion

both = FeatureUnion([("words", TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode")),
                     ("chars", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=3, sublinear_tf=True))])
char_model = make_pipeline(both, LogisticRegression(C=10, max_iter=3000)).fit(train["text"], train["route"])
rows.append(rt.score("TF-IDF words and letters + logistic regression", test, char_model.predict(test["text"])))
rt.show(rows)
```

> **Check.** You should see that the letters do **not** help on this inbox: accuracy 0.962 against 0.968 with words only (French 0.952 against 0.956).

This is a surprise worth keeping. The Larkfield messages were written by a model from short drafts: they have few typos and their words are regular. On a real inbox with many typos and short chat messages, letters often help. The only way to know is to test both on your own messages, as here.

## 6. Failure case: a word the model never saw

TF-IDF knows only the words of the training messages. A new word gets no column at all. The next cell sends a message about a new product, the Grow Hub, with words that the training messages do not have.

```python
new = "My Grow Hub won't pair with the app, the sensor light keeps blinking"
v = word_model[0].transform([new])
print("known words and pairs in the message:", [word_model[0].get_feature_names_out()[i] for i in v.indices])
proba = word_model.predict_proba([new])[0]
print(dict(zip(word_model[-1].classes_, proba.round(2))))
```

> **Check.** Only a few common words are known (`app`, `light`, `my`, `the` …), and no route gets more than half of the probability. The model still has to choose one: here `delivery`, which is wrong.

A new product brings new words. TF-IDF cannot know them until it is trained again with labelled examples. Module 9 shows how to see this happen in production, and Module 4 shows models that know the meaning of words they were not trained on.
