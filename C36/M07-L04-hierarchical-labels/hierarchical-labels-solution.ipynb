## Setup: helpers

The next cell loads pandas as `pd`, numpy as `np` and the course helper as `rt`, reads the three parts of the inbox (`train`, `validation`, `test`), and defines the check helpers. They tell you if your answer is right, without showing the answer:

- `expect(what, yours, expected)` compares one of your numbers with the expected one, and says "too high" or "too low".
- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. The label tree

Each message has a `route` (the team) and a `topic` inside that team. Every topic belongs to exactly one team: the labels form a **tree**.

```python
tree = train.groupby("route")["topic"].unique().apply(sorted)
for route, topics in tree.items():
    print(f"{route:16} {', '.join(topics)}")
print(train["topic"].nunique(), "topics")
```

## 2. The flat model: one model, 27 answers

A **flat** model ignores the tree: it chooses among all 27 topics at once. The team then follows from the topic.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.pipeline import FeatureUnion, make_pipeline

def model():
    return make_pipeline(
        FeatureUnion([("words", TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode")),
                      ("chars", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=3, sublinear_tf=True))]),
        LogisticRegression(C=10, max_iter=3000))

team_of = train.groupby("topic")["route"].first().to_dict()
flat = model().fit(train["text"], train["topic"])
flat_pred = flat.predict(test["text"])
print(f"flat: topic accuracy {(flat_pred == test['topic']).mean():.3f}, macro-F1 {f1_score(test['topic'], flat_pred, average='macro'):.3f}")
```

> **Check.** Accuracy 0.954, macro-F1 0.955. This cell takes about half a minute.

## 3. Your turn: team first, then topic

> **Your turn.** Build the two-step model:
> 1. Train `team_model` on `train["route"]`.
> 2. For each of the 7 routes, train a topic model on the training messages of that route only. Keep them in a dictionary `topic_models`.
> 3. For each test message, take the team that `team_model` predicts, then ask that team's topic model. Put the 2,000 topics in the array `two_step_pred`.

```python
team_model = model().fit(train["text"], train["route"])
topic_models = {r: model().fit(train.loc[train["route"] == r, "text"], train.loc[train["route"] == r, "topic"])
                for r in train["route"].unique()}
teams = team_model.predict(test["text"])
two_step_pred = np.array([topic_models[team].predict([text])[0] for team, text in zip(teams, test["text"])])
print(two_step_pred[:5])
```

> **Check.** Accuracy 0.956, macro-F1 0.958: a little higher than the flat model, by 3 messages in 2,000 for the accuracy.

## 4. Where the two models differ

```python
if two_step_pred is not ...:
    y = test["topic"].to_numpy()
    for name, pred in (("flat", flat_pred), ("two steps", two_step_pred)):
        wrong_team = np.array([team_of[t] for t in pred]) != test["route"].to_numpy()
        print(f"{name:10} right {(pred == y).sum()}, wrong team {wrong_team.sum()}")
    differ = flat_pred != two_step_pred
    print(f"they differ on {differ.sum()} messages: flat right {(differ & (flat_pred == y)).sum()}, "
          f"two steps right {(differ & (two_step_pred == y)).sum()}")
```

> **Check.** Wrong team: flat 80, two steps 76. The models differ on 63 messages: the flat model is right on 21 of them, the two-step model on 24.

The gain is a few messages, inside the noise of a 2,000-message test set. On this inbox, the tree does not help: 27 topics with 44 to 742 training messages each are easy for one flat model. The lesson page says when a tree does help.
