```python
# Helpers: the data, the course helper, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import routing as rt
warnings.filterwarnings("ignore")
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.width", 160)

messages = pd.read_csv("data/messages.csv")
train = messages[messages["split"] == "train"].reset_index(drop=True)
validation = messages[messages["split"] == "validation"].reset_index(drop=True)
test = messages[messages["split"] == "test"].reset_index(drop=True)

def fingerprint(value):
    def norm(v):
        if hasattr(v, "to_dict") and hasattr(v, "columns"):
            return {"columns": [str(c) for c in v.columns], "rows": [norm(list(r)) for r in v.itertuples(index=False)]}
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 4)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. The strong rules and the model

Grace marked four rules as **strong**: the ones that she trusts more than a model. The next cell loads the rule book,
keeps the strong rules, and trains the project's model, TF-IDF + logistic regression, on the training messages.

```python
import json, re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import FeatureUnion, make_pipeline

rules = json.loads(open("data/rules.json", encoding="utf-8").read())["rules"]
for r in rules:
    r["regex"] = re.compile(r["pattern"], re.IGNORECASE)
strong = [r for r in rules if r["strong"]]

def first_match(text, rules):
    for r in rules:
        if r["regex"].search(text):
            return r
    return None

features = FeatureUnion([
    ("words", TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True, strip_accents="unicode")),
    ("chars", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=3, sublinear_tf=True))])
model = make_pipeline(features, LogisticRegression(C=10, max_iter=3000)).fit(train["text"], train["route"])
print([r["why"] for r in strong])
```

> **Check.** You should see the four strong rules: a legal or press threat, a safety problem, someone else used the
> account, and a double charge.

## 2. Three ways to put them together

1. **The model alone.**
2. **Strong rules first:** if a strong rule matches, its route wins; otherwise the model decides.
3. **A disagreement goes to a person:** if a strong rule matches and the model gives another route, a person reads
   the message (`review`); otherwise the model decides.

> **Predict.** Which of the three is the cheapest on the validation messages?

```python
def together(part, how):
    model_pred = model.predict(part["text"])
    out = []
    for text, m in zip(part["text"], model_pred):
        r = first_match(text, strong)
        if r is None or how == "model alone" or r["route"] == m:
            out.append(m)
        elif how == "strong rules first":
            out.append(r["route"])
        else:  # a disagreement goes to a person
            out.append("review")
    return out

ways = ["model alone", "strong rules first", "a disagreement goes to a person"]
rt.show([rt.score(w, validation, together(validation, w)) for w in ways])
```

> **Check.** On validation: the model alone costs 472.0 per 1,000 messages, strong rules first 589.0, and
> a disagreement to a person 432.5.

Strong rules first is the most expensive. Why would rules that Grace trusts make the model worse?

## 3. Each strong rule against the model

For each strong rule, take the validation messages that it matches. On those messages, count how often the rule is
right and how often the model is right.

```python
truth = validation["route"].to_numpy()
model_val = model.predict(validation["text"])
rows = []
for r in strong:
    m = np.array([bool(r["regex"].search(t)) for t in validation["text"]])
    rows.append({"rule": r["why"], "matches": int(m.sum()), "rule right": int((truth[m] == r["route"]).sum()),
                 "model right": int((model_val[m] == truth[m]).sum())})
pd.DataFrame(rows)
```

> **Check.** On the messages that each rule matches, the model is right more often than the rule, for every strong
> rule. For example, the legal-threat rule is right on 16 of 36 messages, the model on 34.

## 4. The disagreements

When a strong rule matches and the model disagrees, who is right?

```python
hits = [first_match(t, strong) for t in validation["text"]]
matched = np.array([h is not None for h in hits])
rule_route = np.array([h["route"] if h else "" for h in hits])
differ = matched & (rule_route != model_val)
print("strong rule matches:", matched.sum(), " disagreements:", differ.sum())
print("model right:", (model_val[differ] == truth[differ]).sum(), " rule right:", (rule_route[differ] == truth[differ]).sum())
```

> **Check.** 159 matches and 70 disagreements. The model is right on 64 of them, the
> rule on 6.

So "strong rules first" changes 70 decisions, and 64 of the changes are wrong. "A disagreement goes to a person"
sends those 70 messages to a review at 1.50 dollars each, and it catches the few expensive mistakes of the model.
