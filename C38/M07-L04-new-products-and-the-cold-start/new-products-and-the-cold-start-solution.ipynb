## Setup: helpers

The next cell loads NumPy as `np`, pandas as `pd`, matplotlib, and the course helper as `cp` (NDCG, recall, MRR, intervals and the comparison table). It defines a few small helpers:

- `catalogue()`, `queries()` and `judgments()` read the data: one row per product, one row per query, and `{query_id: {product_id: grade}}` with grade 2 exact, 1 partial, 0 irrelevant.
- `show(ids)` prints products with their title and class.

It also defines the check helpers. They tell you if your answer is right, without showing the answer:

- `expect(what, yours, expected)` compares one of your numbers with the expected one, and says "too high" or "too low".
- `expect_hash(what, yours, expected)` compares any result with a hidden fingerprint of the expected one.

Run the cell. It prints nothing.

```python
# Helpers: the data, and checks that do not show the answer.
import hashlib, json, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import compare as cp
warnings.filterwarnings("ignore")

def catalogue():
    """One row per product."""
    df = pd.read_csv("data/catalogue.csv.gz", keep_default_na=False, dtype={"title_fr": str})
    for col in ("on_sale", "in_stock"):
        df[col] = df[col].astype(str).eq("True")
    return df

def queries():
    """One row per query with judgments (train, validation and test)."""
    return pd.read_csv("data/queries.csv", keep_default_na=False)

def judgments():
    """{query_id: {product_id: grade}}: 2 exact, 1 partial, 0 irrelevant."""
    j = pd.read_csv("data/judgments.csv.gz")
    out = {}
    for q, p, g in zip(j["query_id"], j["product_id"], j["grade"]):
        out.setdefault(q, {})[p] = int(g)
    return out

def show(ids, judged=None, n=10):
    """Print products: rank, ID, title, class (and the grade when judged is given)."""
    global _CAT
    if "_CAT" not in globals():
        _CAT = catalogue().set_index("product_id")
    cat = _CAT
    names = {2: "exact", 1: "partial", 0: "irrelevant"}
    for i, p in enumerate(list(ids)[:n], 1):
        g = "" if judged is None else f"  [{names.get(judged.get(p), 'not judged')}]"
        print(f"{i:2}. {p}  {cat.loc[p, 'title'][:55]:<55}  {cat.loc[p, 'product_class'][:22]}{g}")

def fingerprint(value):
    def norm(v):
        if hasattr(v, "tolist"):
            return norm(v.tolist())
        if isinstance(v, dict):
            return {str(k): norm(x) for k, x in sorted(v.items(), key=lambda kv: str(kv[0]))}
        if isinstance(v, (list, tuple)):
            return [norm(x) for x in v]
        if isinstance(v, bool) or v is None or isinstance(v, int):
            return v
        if isinstance(v, float):
            return None if v != v else round(v, 3)
        if hasattr(v, "item"):
            return norm(v.item())
        return str(v)
    return hashlib.sha256(json.dumps(norm(value), sort_keys=True).encode()).hexdigest()[:16]

def expect(what, yours, expected, tolerance=1e-6):
    """Compare one number or value with the expected one."""
    if isinstance(yours, (np.integer, np.floating)):
        yours = yours.item()
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
        return
    if isinstance(expected, float) and isinstance(yours, (int, float)):
        same = abs(yours - expected) <= tolerance * max(1, abs(expected))
    else:
        same = yours == expected
    if same:
        print(f"Check passed: {what} is {yours}.")
    elif isinstance(expected, (int, float)) and isinstance(yours, (int, float)):
        print(f"Not yet: {what} is {yours}, which is too {'high' if yours > expected else 'low'}.")
    else:
        print(f"Not yet: {what} is {yours!r}, which is not the expected value.")

def expect_hash(what, yours, expected):
    """Compare any result with the fingerprint of the expected one."""
    if yours is None or yours is Ellipsis:
        print(f"Not yet: {what} has no value. Write your answer in the cell above, then run it again.")
    elif fingerprint(yours) == expected:
        print(f"Check passed: {what} is right.")
    else:
        print(f"Not yet: {what} is not the expected result. Read the task again, and check each step.")
```

## 1. New products have no history

A product added last week has no clicks and no carts in the log. A ranker that learns from popularity puts it low, so nobody sees it, so it gets no clicks. This is the **cold start** problem. The "added" dates are synthetic. Run the next cell.

```python
cat = catalogue().set_index("product_id")
q = queries()
j = judgments()
imp = pd.read_csv("data/log/impressions.csv.gz")      # search_id, position, product_id, clicked, added_to_cart
NEW_FROM = "2025-10-13"                                # the last two weeks of the synthetic log
new = set(cat.index[cat["added"] >= NEW_FROM])
print(len(new), "new products")
print("new products shown at least once:", imp.loc[imp["product_id"].isin(new), "product_id"].nunique())
print("new products clicked at least once:", imp.loc[imp["product_id"].isin(new) & (imp["clicked"] > 0), "product_id"].nunique())
print("all products clicked at least once:", imp.loc[imp["clicked"] > 0, "product_id"].nunique())
```

## 2. Where do the right new products rank?

The judges graded some new products as exact matches for test queries. Where does hybrid search put them?

```python
rr = pd.read_csv("data/recorded/rerank.csv.gz")
test_ids = set(q.loc[q["split"] == "test", "query_id"])
base = {qid: list(g.sort_values("hybrid_rank")["product_id"])
        for qid, g in rr[rr["query_id"].isin(test_ids)].groupby("query_id")}

def new_exact_ranks(run):
    """The rank of every exact new product that the list contains."""
    return [ids.index(p) + 1 for qid, ids in run.items() for p, g in j[qid].items()
            if g == 2 and p in new and p in ids]

r = new_exact_ranks(base)
print(len(r), "exact new products in the lists; median rank", np.median(r))
plt.figure(figsize=(6, 2.4))
plt.hist(r, bins=range(1, 102, 5))
plt.xlabel("rank of an exact new product (hybrid search)"); plt.ylabel("products"); plt.tight_layout(); plt.show()
```

> **Check.** 54 exact new products, with a median rank of 30.5. Most people never scroll that far.

## 3. Your turn: an exploration slot

An **exploration slot** is one fixed place on the page, here position 4, for the best new product of the list. Complete `explore`.

```python
def explore(ids, slot=4):
    """The best new product of the list goes to position `slot` (1 = first), if it is lower than that."""
    fresh = [p for p in ids if p in new]
    if not fresh or ids.index(fresh[0]) < slot:
        return ids
    rest = [p for p in ids if p != fresh[0]]
    return rest[:slot - 1] + [fresh[0]] + rest[slot - 1:]

run = {qid: explore(ids) for qid, ids in base.items()}
median_after = float(np.median(new_exact_ranks(run)))
print("median rank of exact new products:", median_after)
```

```python
expect('median_after', median_after, 4.0)
```

> **Check.** The median rank of the exact new products is now 4.0.

## 4. What the slot costs

The slot also moves new products that are not right into position 4, and it pushes one product down on every page.

```python
if median_after is not None:   # after your explore() works
    d = cp.paired(cp.per_query(base, j), cp.per_query(run, j))
    print(f"exploration slot minus hybrid: NDCG@10 {d['difference']:+.3f} "
          f"(95% interval {d['low']:+.3f} to {d['high']:+.3f}); better on {d['better']}, worse on {d['worse']}")
    shown = sum(any(p in new for p in ids[:10]) for ids in run.values())
    print("test queries with a new product in the top 10:", shown, "of", len(run))
```

> **Check.** NDCG@10 changes by -0.013 on the test queries, and 193 queries now show a new product in the top 10. The slot buys exposure with a little relevance: the clicks it earns are what the ranker needs next time.
